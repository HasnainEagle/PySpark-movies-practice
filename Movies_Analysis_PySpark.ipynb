### Movies EDA Using Pyspark(Practice)

```python
# reading dataset
movies_df = spark.sql('select * from workspace.default.movies')
movies_df.limit(5).display()
```

```python
from pyspark.sql.functions import *
# drop unnecessary columns 
movies_df = movies_df.drop(col('Poster_Link'))
movies_df.columns
```

```python
# Create a new column that shows the difference between Meta_score and IMDB_Rating.
movies_df.withColumn('Difference',col('Meta_score')-col('IMDB_Rating')).display()
```

```python
# Clean the Runtime column so it only contains numbers in minutes.  
# movies_df.select('Runtime').distinct().display()
movies_df = movies_df.withColumn('Runtime',split(col('Runtime'),' ')[0].cast('int'))
```

```python
# Breakown the Genre column so each genre has its own row, even if a movie has multiple genres.
# movies_df.select('Genre').display()
movies_df = movies_df.withColumn('Genre',split(col('Genre'),', '))
movies_df = movies_df.withColumn('Genre',explode('Genre'))
movies_df.select('Series_Title','Genre').display()
```

```python
# Standardize the Released_Year column by removing extra characters or missing values.
# movies_df.select('Released_Year').distinct().display()
# movies_df.groupBy('Released_Year').count().orderBy(col('count').desc()).display()
movies_df = movies_df.withColumn('Released_Year',when(col('Released_Year').rlike('^[0-9]{4}$'),col('Released_Year').cast('int')).otherwise(None))
# movies_df.groupBy('Released_Year').count().orderBy(col('count').desc()).display()
movies_df = movies_df.withColumn('Released_Year',trim(col("Released_Year")).cast('int'))
movies_df.schema['Released_Year'].dataType
```

```python
# Convert the Gross column into numeric format and handle missing or corrupted values.
movies_df = movies_df.withColumn('Gross',regexp_replace(col('Gross'),",",""))
movies_df = movies_df.fillna({'Gross':'0'})
movies_df = movies_df.withColumn('Gross',col('Gross').cast('int'))
movies_df.groupBy('Gross').count().orderBy('count',ascending = False).display()
movies_df.schema['Gross'].dataType
# movies_df.filter(col('Gross').isNull()).count()
```

```python
# Find the top 3 movies by rating in each genre.
from pyspark.sql.window import Window
window = Window.partitionBy('Genre').orderBy(col('IMDB_Rating').desc())
top_3 = movies_df.withColumn('rank',dense_rank().over(window)).filter(col('rank') <= 3)
top_3.select('Series_Title','Genre','IMDB_Rating','rank').display()
```

```python
# Find the top 5 highest-grossing movies for each year.
window2 = Window.partitionBy('Released_Year').orderBy(col('Gross').desc())
top_five = movies_df.withColumn('rank',dense_rank().over(window2)).filter(col('rank') <= 5)
top_five.select('Series_Title','Released_Year','Gross','rank').display()
```

```python
# List the movies with the highest and lowest votes for every year
window3 = Window.partitionBy('Released_Year')
res = movies_df.withColumn('max_votes',max(col("No_of_Votes")).over(window3))\
         .withColumn('min_votes',min(col("No_of_Votes")).over(window3))
res.select('Series_Title','Released_Year','No_of_Votes','max_votes','min_votes').display()
```

```python
# Find all movies where IMDB_Rating is above the overall average rating.
overall_avg_rating = movies_df.select(mean(col('IMDB_Rating'))).first()[0]
movies_df.filter(col('IMDB_Rating') > overall_avg_rating).select('Series_Title','IMDB_Rating').display()
```

```python
# Find directors who have directed more than 5 movies and their average gross earnings.
movies_df.groupBy('Director').agg(mean('Gross').alias('average_gross'),count("*").alias('count')).filter(col('count') > 5).display()
```

```python
# Create a column that categorizes movies as Short (<90 min), Medium (90–150 min), or Long (>150 min).
movies_df.withColumn('Categroization_Runtime',when(col('Runtime') < 90,'Short')\
                                             .when((col("Runtime") >= 90) & (col("Runtime") <= 90),'Medium')\
                                             .otherwise('Long')).display()
```

```python
# For each year, find the total votes cast for all movies released in that year.
movies_df.groupBy('Released_Year').agg(sum("No_of_Votes").alias('Total_No_Votes')).orderBy(col('Total_No_Votes').desc()).display()
```

```python
# For each genre, find the oldest and newest movie released.

Genre_window = Window.partitionBy('Genre')

int_df = movies_df.withColumn('Oldest_Year',min('Released_Year').over(Genre_window))\
         .withColumn('Newest_Year',max('Released_Year').over(Genre_window))

oldest_movie = int_df.filter(col('Released_Year') == col('Oldest_Year')).select('Genre','Released_Year',col('Series_Title').alias('Oldest_Movie'))

newest_movie = int_df.filter(col('Released_Year') == col('Newest_Year')).select('Genre','Released_Year',col('Series_Title').alias('Newest_Movie'))

combined = oldest_movie.join(newest_movie,
                  on = [oldest_movie.Genre == newest_movie.Genre],
                  how = 'inner').select(oldest_movie.Genre.alias("Genre"),'Oldest_Movie','Newest_Movie')
# oldest_movie.Genre.alias("Genre") ---> to avoid ambiguous error
combined.display()
```

```python
# Find movies that rank in the top 10 for both IMDB_Rating and Meta_score.
movies_df\
    .withColumn('rank',dense_rank().over(Window.orderBy(col('IMDB_Rating').desc(),col('Meta_score').desc())))\
    .filter(col('rank') <= 10).display()
```

```python
# List all movies that earned more than the average gross of their release year.

# window
year_window = Window.partitionBy('Released_Year')

result = movies_df.withColumn('Average_Gross_Per_Year',mean(col('Gross')).over(year_window))\
         .filter(col('Gross') > col('Average_Gross_Per_Year'))\
         .select('Series_Title','Gross','Released_Year','Average_Gross_Per_Year').distinct()
result.display()
```

```python
# Group movies by certificate and compute their average runtime and average rating.
movies_df.groupBy('Certificate').agg(mean('Runtime').alias('average_runtime'),mean('IMDB_Rating').alias('average_rating')).display()
```

```python
# Combine the four star columns into a single array column of actors.
movies_df.withColumn('actors',array(col('Star1'),col('Star2'),col('Star3'),col('Star4'))).display()
```

```python
# Rank genres based on total votes received across all movies.

movies_df.groupBy('Genre').agg(sum('No_of_Votes').alias('Total_No_Of_Votes')).orderBy(col('Total_No_Of_Votes').desc()).display()
```

```python
# Identify the most common certificate for movies in each genre.
gen_cer_count = movies_df.groupBy('Genre','Certificate').agg(count("*").alias('count'))

window_genre = Window.partitionBy('Genre').orderBy(col('count').desc())
ranked = gen_cer_count.withColumn('rank',dense_rank().over(window_genre))
res = ranked.filter(col('rank') == 1)
res.display()
```

```python
# Find the top 10 movies with the highest ratio of Gross to Runtime.
res = movies_df.drop_duplicates(subset=['Series_Title'])
res.withColumn('Gross_to_runtime_ratio',(col('Gross')/col('Runtime'))).orderBy(col('Gross_to_runtime_ratio').desc()).limit(10).select('Series_Title','Gross_to_runtime_ratio').display()
```

```python
# Find the director who has the highest-grossing movie in each decade.
decade = movies_df.withColumn('Decade',(floor(col('Released_Year')/10)*10))
# decade.select('Released_Year','Decade').display()
decade_window = Window.partitionBy('Decade').orderBy(col('Gross').desc())
decade.withColumn('rank',dense_rank().over(decade_window)).filter(col('rank') == 1).select('Decade','Director','Gross').distinct().display()
```

```python
# Identify which actor (from Star1–Star4) appears most often in the dataset.

# My Approach:
# look at my aproach : first i group by star1.count().alias(count_1), idid the same for all the stars columns after this i have 4 dataframe name star1_df,star2_df and so on and after this i will full join star1 and star 2 ---> join1 ,star3 and star4 ---> join 2 then i full join join1 and join2 and then group by star and sum count

# efficient_approach
all_actors = movies_df.withColumn('All_Stars',array(col('Star1'),col('Star2'),col('Star3'),col('Star4')))
all_actors = all_actors.withColumn('All_Stars',explode('All_Stars'))
common_actor = all_actors.groupBy('All_Stars').agg(count("*").alias('count')).orderBy(col('count').desc()).limit(1)
common_actor.select('All_Stars').first()[0]
```

```python
# Find movies where the same actor worked with the same director more than once.

res = movies_df.drop_duplicates(subset=['Series_Title'])
all_actors = res.withColumn('actors',array(col('Star1'),col('Star2'),col('Star3'),col('Star4')))
actor_per_row = all_actors.withColumn('actors',explode('actors'))
actor_director_count = actor_per_row.groupBy('actors','Director').count().orderBy(col('count').desc())
actor_director_count.filter(col('count') > 1).display()
```

```python
# Find movies released in consecutive years by the same director.

res = movies_df.drop_duplicates(subset=['Series_Title'])
res = res.filter(col('Released_Year').isNotNull())

# define window
director_window = Window.partitionBy('Director').orderBy('Released_Year')

# use lag to find prev_year
year_director = res.withColumn('prev_year',lag('Released_Year',1).over(director_window))

# compute diff
compute_movies = year_director.filter((col('prev_year').isNotNull()) & ((col('Released_Year') - col('prev_year')) == 1).alias('diff'))
compute_movies.select('prev_year','Released_Year','Director','Series_Title').display()
```

```python
# Identify movies where the overview length (number of words) is unusually short or long.
res = movies_df.drop_duplicates(subset=['Series_Title'])
x = res.withColumn('overview_length',when((size(split(col('Overview'),' '))) <= 20 , 'Short')\
                                    .when((size(split(col('Overview'),' '))) >= 100 , 'Long')\
                                    .otherwise('Normal'))
x.select('Series_Title','overview_length').display()
```

```python
# Create a column that calculates revenue per vote (Gross / No_of_votes).
res = movies_df.drop_duplicates(subset=['Series_Title'])
res.withColumn('revenue_per_vote',col('Gross')/col('No_of_Votes')).display()
```

```python
# Identify movies that have the same Series_Title but were released in different years.
res = movies_df.drop_duplicates(subset=['Series_Title'])
a_df = res.alias('a')
b_df = res.alias('a')
combined_res = a_df.join(
    b_df,
    ((a_df.Series_Title == b_df.Series_Title) & (a_df.Released_Year != b_df.Released_Year))
).select(a_df.Series_Title,a_df.Released_Year,b_df.Series_Title,b_df.Released_Year)
combined_res.display()
```

```python
# List all movies where at least two of the same stars appear together in another movie.
res = movies_df.drop_duplicates(subset=['Series_Title'])

# combined all the stars cols into an array
res = res.withColumn('actors',array(col('Star1'),col('Star2'),col('Star3'),col('Star4')))

# then explode so each row has only one star
res = res.withColumn('actor',explode('actors'))

# alias dataframe
first = res.alias('first')
second = res.alias('second')

# join logic
joined_df = first.join(
    second,
    ((first.Series_Title != second.Series_Title) & (first.actor == second.actor))
)

# group by movies and count the actors appearence
result = joined_df.groupBy(first.Series_Title,second.Series_Title)\
         .agg(collect_set(first.actor).alias('shared_actors'))\
         .withColumn('shared_actors_count',size('shared_actors'))\
         .filter(col('shared_actors_count') >= 2)
result.select(first.Series_Title,second.Series_Title,col('shared_actors')).display()
```

```python
# Create a running ranking of movies based on votes within each year.

running_window = Window.partitionBy('Released_Year').orderBy('no_of_votes')

res = movies_df.drop_duplicates(subset=['Series_Title'])
running_ranking = res.withColumn('running_rank',dense_rank().over(running_window))
running_ranking.select('Released_Year','Series_Title','no_of_votes','running_rank').display()
```
